### 패키지(Package)

파이썬의 패키지는 디렉토리와 모듈(~.py)로 구성된다.

1. 물리적으로 여러개의 모듈파일을 모아 놓은 디렉토리이다.
2. 모듈은 `~.py`인 파일이고, 패키지는 `디렉토리 즉 폴더`이다.
3. `__init__.py`(__init__메서드 아님)은 파일을 해당 디렉토리가 패키지임을 파이썬에 엔인에게 알려주는 역할을 한다
    - 파이썬 v3.3 이후 부터는 이 파일이 없어도 해당 디렉토리가 패키지로 인식이 된다.
    - 파이썬 이전 버전에서는 디렉토리 안에 **__init__.py** 파일이 없으면 패키지로 인식되지 않는다
    - 하위버전과 호환성을 위해서라도 dummy파일일지라도 이 파일을 생성하는것이 좋다.
4. 파이썬 모듈은 계층적(디렉토리 구조)으로 관리하기 위해서는 `dot연산자`를 사용한다. 예를 들어 모듈명이 "project.source.mod2"일 경우에는 패키지명은 project.sourc이고, 모듈명은 전체가 모듈명이 된다.
