## 튜플(tuple) 자료형
1. 튜플자료형은 수정이 불가능하며, 순서가 있는 `객체의 집합`이다.
2. 튜플은 리스트자료형과 비슷하지만, 한번 생성이 됨ㄴ 값을 변경할 수 없다, 즉, list는 수정 삭제, 추가가 가능하지만, tuple은 수정, 삭제, 추가가 불가능한 `immutable 자료형`이다.
3. list는 대괄호[]로 정의 하지만 tuple은 `소괄호()`로 정의한다.
4. 순서가 있기 때문에 index로 접근이 가능하다
5. tuple은 list와 같이 `더하기(+), 반복하기(*)` 연산을 할 수가 있다
6. tuple은 다양한 자료형을 요소로 가질 수 있다.
7. `소괄호()안에 1개의 요소만 있을 경우에는 tuple자료형이 아니라 요소값의 자료형`으로 되지만 1개의 요소에 콤마(,)를 찍어주면 tuple로 선언된다.
8. 함수에 값을 리턴할 때 `tuple을 이용하면 변수를 일괄로 할당`할 수 있다.


1. tuple 선언하기


```python
# 1. tuple의 초기화 = 빈 tuple 객체를 선언

l = []
l = list()
print(type(l))

t = ()
t = tuple()
print(type(t))
```
