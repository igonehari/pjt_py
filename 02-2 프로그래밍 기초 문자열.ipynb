#### 2. 문자형(문자열)
* 파이썬에서 문자열(str)이란 문자열, 즉 문자의 집합을 의미한다.
* 문자열은 작은.큰 따옴표 구분없이 사용
* 여러줄은 문자열을 상의할 때는 큰 따옴표 3개 즉, """로 감싸서 정의한다

```python
food = "Python's favorite food is perl"
food
```

```python
food = 'Python's favorite food is perl
```

```python
say = "Python is very easy." he says.
```

```python
say = '"Python is very easy." he says.'
say
```

문자열 포매팅 따라하기

1. 숫자 바로 대입

```python
"I eat %d apples."% 3
```

2. 문자열 바로 대입

```python
"I eat %s apples." % "five"
```

3. 숫자 값을 나타내는 변수로 대입

```python
number = 3
"I eat %d apples." % number
```

4. 2개 이상의 값 넣기

```python
number = 10
day ="three"
"I ate %d apples. so I was sick for %s days" % (number, day)
```
