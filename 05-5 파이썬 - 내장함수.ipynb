#### abs
abs(x)는 어떤 숫자를 입력 받았을 때 그 숫자의 절댓값을 리턴하는 함수이다.

```python
print(abs(3))
print(abs(-3))
print(abs(-1.2))
```

#### all
all(x)는 반복 가능한 데이터 x를 입력값으로 받으며 이 x의 요소가 모두 참이면 True, 거짓이 하나라도 있으면 False를 리턴한다.

```python
all([1, 2, 3]) # 모두 참
```

```python
all([1, 2, 3, 0]) # 0은 거짓
```

```python
all([]) # 인수가 빈 값인 경우에는 True
```

#### any
any(x)는 반복 가능한 데이터 x를 입력으로 받아 x의 요소 중 하나라도 참이 있으면 True를 리턴하고 x가 모두 거짓일때만 False를 리턴한다, 즉 all의 반대

```python
any([1, 2, 3, 0])
```

```python
any([0, ""])
```

```python
any([]) # any도 인수가 빈 값일 경우에는 False 리턴
```

#### chr
chr(i)는 유니코드 숫자값을 입력받아 그 코드에 해당하는 문자를 리턴하는 함수이다.

```python
print(chr(97))
print(chr(44032))
```
