예제를 통해 for 문 이해하기

```python
test_list = ['one', 'two', 'three']
for i in test_list:
    print(i)
```

다양한 for 문의 사용

```python
a = [(1, 2), (3, 4), (5, 6)]
for (first, last) in a:
    print(first + last)
```

for 와 range를 이용한 구구단

```python
for i in range(2, 10):
    for j in range(1, 10):
        print(i*j, end=" ")
    print('')
```

리스트 컴프리헨션 사용하기

```python
a = [1, 2, 3, 4]
result = []
for num in a:
    result.append(num*3)

print(result)
```

더 간단하게 만들기

```python
a = [1, 2, 3, 4]
result = [num*3 for num in a]
print(result)
```

짝수만 곱하기

```python
a = [1, 2, 3, 4]
result = [num*3 for num in a if num%2 ==0]
print(result)
```

[표현식 for 항목 in 반복_가능_객체 if 조건문]


```python
result = [x*y for x in range(2, 10)
          for y in range(1, 10)]

print(result)
```

```python
for i in range(1, 6):
    print("*"*i)
```

```python
a = 5
for i in range(0, a):
    print(" "*(a-1-i) + "*"+ "**"*i)
```

```python
for i in range(1, 10, 2):
    print(f'{"*"*i:^9}')
```

```python
a = 5
for i in range(0, a):
    print(" "*(a-1-i) + "*"+ "**"*i)
for i in range(1, a):
    print(" "*(i) + "*"+ "**"*(a-1-i))
```

```python
import turtle as t

t.shape("turtle")
t.color("red")
t.begin_fill()

for _ in range(5):
    t.forward(150)
    t.right(144)

t.end_fill()
t.done()
```
