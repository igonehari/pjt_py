### 집합자료형(set)

* 집합자료형은 파이썬 v2.3부터 지원하기 시작한 자료형이다.
* 이 자료형은 경험과 관련된 작업을 보다 쉬게 처리할 수 있도록 만들어진 자료형이다.
* 리스트나 튜플은 순서가 있기 때문에 인덱스를 통해서 값을 가져올 수 있지만, set자료형은 `순서와 ket가 없기` 때문에 인덱스나 key로 값을 가져올 수 없다.
* 따라서, set자료형을 값을 가져오기 위해서는 자료형을 `list or typle 자료형으로 변환 후에 인덱스로 값을 가져올 수 있다.`
* set 자료형은 `집합관련 연산`이나 `중복을 제거하기 위한 필터` 역할로 종종 사용한다.
* set자료형으로 선언하려면 `중괄호안에 값을 가진 객체로 선언`하거나 `set()ㅣ라는 생성자`로 선언한다.
* 생성자 set()이 iterable한 객체를 전달하면 자동으로 set자료형으로 변환해준다.

#### 집합자료형의 특징
1. `중복을 허용하지 않고 집합안에서 unique 한 값`을 가진다.
2. `순서가 없다`
3. set자료형의 선언은 `중괄호{} or set()`로 선언한다.
4. 중괄호로 선언하는 것은 dict와 유사하지만 set자료형은 key가 없고, `값만 존재`한다.
5. set자료형의 내부요소 즉 값은 다양한 값을 가질 수 있지만 `mutable 한 자료는 가질 수 없다.`

집합 자료형은 어떻게 만들까

```python
s1 = set([1,2,3])
s1
```

```python
s2 = set("Hello")
s2
```

```python
s1 = set([1,2,3])
l1 = list(s1)
l1
l1[0]
t1 = tuple(s1)
t1
t1[0]
```

교집합 구하기

```python
s1 = set([1,2,3,4,5,6])
s2 = set([4,5,6,7,8,9])
s1 & s2
s1.intersection(s2)
```

합집합 구하기

```python
s1 | s2
```

```python
s1.union(s2)
```

차집합 구하기

```python
s1 - s2
s2 - s1
s1.difference(s2)
s2.difference(s1)
```

값 1개 추가하기

```python
s1 = set([1,2,3])
s1.add(4)
s1
```

값 여러 개 추가하기

```python
s1 = set([1,2,3,])
s1.update([4,5,6])
s1
```

특정 값 제거하기

```python
s1 = set([1,2,3])
s1.remove(2)
s1
```
