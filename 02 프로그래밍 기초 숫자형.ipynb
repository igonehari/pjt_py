A. 파이썬 개요
#### 1. 파이썬 개요

1. 1989년 12월에 네덜란드 출신 개발자 **귀도 반 로섬**이 성탄절을 의미있게 보내기 위해 만든 인터프린터 프로그램 언어이다.
2. 파이썬 : 자신이 좋아하는 코미디쇼 "몬티 파이썬의 날아다니는 써커스"에서 따왔음
3. 파이썬 : 사전적인 의미 : 파르나소스산 동굴에 살던 큰 뱀
4. 파이썬 사용되는 대표적인 사례
    * 구글 : c++, Java, Python을 사용, 구글의 내부시스템과 많은 구글 API들이 파이썬으로 작성
    * 유튜브
    * 드롭박스
    * 페이스북
    * 인스타그램

#### 2. 파이썬의 라이버러리(패키지)
1. `Numpy` : 수치계산용 라이브러리(https://numpy.org)
2. `Pandas` : 데이터분석용 라이브러리(https://pandas.pydata.org)
3. `matplotlib` : 시각화(그래프출력용) 라이브러리(https://matplotlib.org)
4. 기타 : `scipy, sklearn, tensortflow, keras`

1. 파이썬은 인간다운 언어

#### 2. 주석
* 파이썬은 한 줄 주석 : #
* 여러 줄 주석 : `""" """` or `''' ''''`
* 단축키 : `ctrl + /`로 주석/해제

```python
# 주석입니다.

"""
여러줄의
주석을
삽입합니다.
"""

'''
여러줄의
주석을
작은따옴표로
삽입니다

'''
print("Hello")
print('Hello')
```
