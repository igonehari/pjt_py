## 리스트(list) 자료형
* 파이썬에서는 리스트 자료형을 만들 때 `대괄호[]`으로 감싸주고 그 안에 요소들을 `콤마(,)`로 구분해서 입력
* 파이써에서는 리시트자료형은 String(str, 문자열)만큼 자주 사용되는 자료형임
> `변수명 = [요소1, 요소2, .... 요소n]`

#### 리스트(list)란?
1. 순서가 있는 자료형으로서 수정이 가능한 `객체의 집합`이다.
2. 수정, 삭제 추가가 가능하다
3. list는 `대괄호[]`로 선언하고, 내부요소는 콤마로 구분한다
4. list의 `인덱스는 0부터 시작`된다.
5. index의 범위가 벗어나는 경우에는 에러가 발생된다.
6. list는 어떠한 자료형오 함께 선언할 수 있다. 즉, 숫자, 문자, 함수, 리스트, 튜플, 문자열 등의 모든 자료형을 요소로 정의할 수 있다.
7. list는 `마지막 요소뒤에 콤마(,)`가 있어도 에러는 발생되지 않는다. 보통 생략을 하지만 편의를 위해서 콤마(,)를 추가하기도 한다.

```python
# 슬라이싱
# 문자열과 마찬가지로 리스트에서도 슬라이싱 기법을 사용할 수 있다
# 파이썬에서는 리스트는 문법이 문자열에서 했던 방법과 동일
# [start:end:step]

l = [1, 2, 3, 4, 5]

print(l[:])
print(l[::2])
print(l[::3])
print()
print(l[1:3])
s = '12345'
print(s[1:3])
print()
print(s[:2])
print(s[2:])
print(s[2::2])
```
