### 변수
* 변수란? `자료형의 값을 저장하는 메모리 공간`을 말한다.
* 파이써에서 변수를 만드는 방법은 `변수명 = 값` 즉, `대입연산자(assignment), =`을 사용한다.
* C, Java처럼 변수의 자료형과 함께 사용하지 않는다.
* 파이썬에서는 변수에 저장한 값을 가지고 스스로 판단해서 자료형을 알아낸다.
* 파이썬에 사용하는 `변수는 객체(object)의 메모리주소를 가지고 있다.`즉, `객체가 저장되어 있는 메모리 주소`를 가리키는 것이다.
* 파이썬에서는 숫자(정수) 3은 일반적으로 상수값이지만, 파이선에서는 `3은 정수형 객체(object)`이다.
* 예를 들어서 `print(type(3))`의 결과는 `<class int>`이다. 즉, `은 int클래스의 객체`임을 알 수 있다.

#### 2. 변수의 명명규칙
1. 영문자와 숫자를 사용할 수 있다.
2. 대소문자를 구분해야 한다.
3. 문자부터 시작해야 하고, 숫자로 시작하면 안된다.
4. `언더바(_)`로 시작할 수 있다.
5. 특수문자(+, -, *, /등)는 사용할 수 없다.
6. 파이썬의 키워드(if, for, def...등)는 사용할 수 없다.

변수를 복사하는 방법

```python
# 얖은 복사 (메모리 주소가 같다)
l1 = [1,2,3]
l2 = l1

#깊은 복사 (메모리 주소가 다르다)
l3 =l1[:]

#깊은 복사
l4 = l1.copy()

print(l1, l2, l3, l4)
```

## 증감, 할당 연산자

* =  : 왼쪽변수에 오른쪽 값을 항당
* += : 왼쪽변수에 오른쪽 값을 더한 후, 그 결과를 왼쪽 변수에 할당
* -= : 왼쪽변수에 오른쪽 값을 뺀 후, 그 결과를 왼쪽 변수에 할당
* *= : 왼쪽변수에 오른쪽 값을 곱한 후, 그 결과를 왼쪽 변수에 할당
* /= : 왼쪽변수에 오른쪽 값을 나눈 후, 그 결과를 왼쪽 변수에 할당
* %= : 왼쪽변수에 오른쪽 값을 나눈 후, 그 나머지를 왼쪽 변수에 할당


[:]이용하기

```python
a = [1,2,3]
b = a[:]
a[1] = 4
a
b
```

copy 모듈 이용하기

```python
from copy import copy
a = [1,2,3]
b = copy(a)
b is a
```

변수를 만드는 여러가지 방법

```python
a, b = ('python', 'life')
(a, b) = 'python', 'life'
[a, b] = ['python', 'life']
a = b = 'python'
```
