## 딕셔너리(Dictionary) 자료형

<img src="./images/04.파이썬 자료형_딕셔너리_01.png" with = "500" height = "400">

* 파이썬에서의 딕셔너리자료형이란 `key와 value를 한쌍`으로 갖는 자료형을 말한다.
* 파이썬에서는 이러한 자료형을 `Dictionary`라고 하는데 딕셔너리는 튜플이나 리스트처럼 순차적(sequential)으로 즉, index로 해당 요소값을 처리하지 않고,`key를 통해서 value에 접근`한다.

#### 딕셔너리의 특징
1. 딕셔너리의 선은 `중괄호{} or dict()`로 한다.
>{key1:value1, key2:value2.....}
2. 딕셔너리는 `immutable한 key`와 `mutable한 value`로 매핑되어 있는 순서가 없는 자료형이다.
3. 이런 이유로 key값으로는 immutable한 값을 사용할 수 있지만, mutable한 객체는 사용할 수 없다.
* 튜플은 key로 사용가능, 리스는 불가능
4. value는 중복이 될 수 있지만, key는 중복이 될 수 없다. key가 중복이 된다며 마지막인 key와 value로 결정된다.
5. 순서가 없기 때문에 인덱스로 접근할 수 없고, `key로 value에 접근` 할 수 있다.
6. value는 mutable한 객체이기 때문에 `key로 접근해서 value를 수정할 수 있다.`

딕셔너리는 어떻게 만들까?

```python
dic = {'name' : 'pey', 'phone' : '010-9999-1234', 'birth': '1118'}
```

딕셔너리 쌍 추가하기

```python
a = {1: 'a'}
a[2] = 'b'
a
```

```python
a['name'] = 'pey'
a
```

```python
a[3] = [1, 2, 3]
a
```

딕셔너리 요소 삭제하기

```python
del a[1]
a
```

딕셔너리에서 Key를 사용해 Value 얻기

```python
grade = {'pey' : 10, 'julliet' : 99}
grade['pey']
grade['julliet']
```

```python
a = {1: 'a', 2: 'b'}
a[1]
a[2]
```

```python
dic = {'name' : 'pey', 'phone' : '010-9999-1234', 'birth' : '1118'}
dic['name']
dic['phone']
dic['birth']
```
