#### 불(Bool, Boolean) 자료형

* bool 자료형이란 `참과 거짓을 나타내는 자료형`이다.
* boot 자료형은 2가지 값인 `True` or `False`값만 가질 수 있다.
* 주의할 점은 True와 False는 파이썬의 예약어로서 첫글자는 반드시 대문자여야 한다.

불 자료형은 어떻게 사용할까?

```python
a = True
b = False
```

```python
type(a)
type(b)
```

```python
1 == 1
```

```python
2 > 1
```

```python
2 < 1
```

논리 연산자
* and
* or
* not

* and 연산자는 양쪽 조건이 모두 참일때만 True를 반환함
* or 연산자는 양쪽 조건 중 하나라도 참이면 True를 반환함
* not 연산자는 참과 거짓을 반대로 바꾼다

자료형의 참과 거짓

```python
# "Python"    참
# ""          거짓
# [1,2,3]     참
# []          거짓
# (1,2,3)     참
# ()          거짓
# {'a':1}     참
# {}          거짓
# 1           참
# 0           거짓
# None        거짓
```

```python
a = [1, 2, 3, 4]
while a:
    print(a.pop())
```

```python
if []:
    print("참")
else:
    print("거짓")
```

```python
if [1,2,3]:
    print("참")
else:
    print("거짓")
```
